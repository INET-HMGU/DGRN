```
# Example: Extract promoter sequences from genome FASTA and GFF annotation
# This code assumes you have 'genome.fasta' and 'genes.gff' files available

from Bio import SeqIO
import pandas as pd

def extract_promoters_from_gff(genome_fasta, gff_file, promoter_length=1000, feature_type='gene'):
    # Load genome sequence
    genome = SeqIO.to_dict(SeqIO.parse(genome_fasta, 'fasta'))
    
    # Load GFF annotation
    gff_cols = ['seqid', 'source', 'type', 'start', 'end', 'score', 'strand', 'phase', 'attributes']
    gff = pd.read_csv(gff_file, sep='\t', comment='#', names=gff_cols)
    
    # Filter for gene features
    genes = gff[gff['type'] == feature_type]
    
    # Prepare results
    promoter_records = []
    
    for idx, row in genes.iterrows():
        chrom = row['seqid']
        strand = row['strand']
        start = int(row['start'])
        end = int(row['end'])
        attrs = row['attributes']
        gene_id = attrs.split(';')[0].replace('ID=', '') if 'ID=' in attrs else attrs
        
        if strand == '+':
            promoter_start = max(0, start - promoter_length - 1)
            promoter_end = start - 1
            promoter_seq = genome[chrom].seq[promoter_start:promoter_end]
        else:
            promoter_start = end
            promoter_end = min(len(genome[chrom]), end + promoter_length)
            promoter_seq = genome[chrom].seq[promoter_start:promoter_end].reverse_complement()
        
        promoter_records.append({
            'gene_id': gene_id,
            'chrom': chrom,
            'strand': strand,
            'promoter_start': promoter_start + 1,
            'promoter_end': promoter_end,
            'promoter_seq': str(promoter_seq)
        })
    
    promoters_df = pd.DataFrame(promoter_records)
    return promoters_df

# Example usage (uncomment and set your file paths):
# promoters = extract_promoters_from_gff('genome.fasta', 'genes.gff', promoter_length=1000)
# print(promoters.head())
```

```
# Parse the Eutrema salsugineum GFF3 annotation file to get gene coordinates and strand information
import re

gff_file = 'Eutrema_ensembl_genomic.gff3'
gff_cols = ['seqid', 'source', 'type', 'start', 'end', 'score', 'strand', 'phase', 'attributes']
eutrema_gff = pd.read_csv(gff_file, sep='\t', comment='#', names=gff_cols)

# Filter for CDS features (to get ATG start codons)
eutrema_cds = eutrema_gff[eutrema_gff['type'] == 'CDS']

# Extract gene_id from attributes
eutrema_cds['gene_id'] = eutrema_cds['attributes'].str.extract('Parent=([^;]+)')

# Get the first CDS for each gene (ATG)
eutrema_cds_atg = eutrema_cds.sort_values(['gene_id', 'start']).groupby('gene_id').first().reset_index()

print('Eutrema CDS ATG data:')
print(eutrema_cds_atg[['seqid', 'gene_id', 'start', 'end', 'strand']].head())
```

```
# Extract -1300 to +300 sequences relative to ATG for Eutrema salsugineum
from Bio.Seq import Seq

eutrema_records = []
for idx, row in tqdm(eutrema_cds_atg.iterrows(), total=eutrema_cds_atg.shape[0]):
    chrom = str(row['seqid'])
    gene_id = row['gene_id']
    atg_start = int(row['start']) if row['strand'] == '+' else int(row['end'])
    strand = row['strand']
    
    if chrom not in eutrema_genome:
        continue
    seq = eutrema_genome[chrom].seq
    if strand == '+':
        seq_start = max(0, atg_start - 1300 - 1)
        seq_end = atg_start + 300 - 1
        subseq = seq[seq_start:seq_end]
    else:
        seq_start = max(0, atg_start - 300 - 1)
        seq_end = atg_start + 1300 - 1
        subseq = seq[seq_start:seq_end].reverse_complement()
    eutrema_records.append({
        'gene_id': gene_id,
        'chrom': chrom,
        'strand': strand,
        'atg': atg_start,
        'seq': str(subseq)
    })

eutrema_promoter_df = pd.DataFrame(eutrema_records)
print('Eutrema promoter sequences extracted: ' + str(len(eutrema_promoter_df)))

# Save to FASTA file
with open('Eutrema_promoters_-1300+300.fa', 'w') as fasta_out:
    for i, row in eutrema_promoter_df.iterrows():
        fasta_out.write('>' + row['gene_id'] + ' ' + row['chrom'] + ' ' + row['strand'] + ' ATG:' + str(row['atg']) + '\n')
        fasta_out.write(row['seq'] + '\n')

print('Eutrema promoter sequences saved to Eutrema_promoters_-1300+300.fa')
```
